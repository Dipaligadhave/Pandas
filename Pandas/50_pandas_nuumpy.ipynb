Q: Why use rolling computations?

A: They help analyze trends over time by smoothing short-term fluctuations.

Examples:

Stock market moving averages
Daily temperature averages
Weekly sales trends
Sensor data smoothing

A rolling window looks at a small chunk of recent values and computes a statistic such as mean, sum, min, max, or standard deviation.

```python
### NP49. Perform a mini data analysis project.
np.random.seed(42)
n=100
student=pd.DataFrame({
    'student_id' : np.arange(1,n+1),
    'subjects' : np.random.choice(['python','java','c++','c']),
    'python':  np.random.randint(40 , 101 ,n).astype(float),
    'java':    np.random.randint(40 , 101 ,n).astype(float),
    'c++':     np.random.randint(40 , 101 ,n).astype(float),
    'c':       np.random.randint(40 , 101 ,n).astype(float),
})
print(student)
student.loc[67, 'python'] = np.nan
student.loc[50, 'java'] = np.nan
student.loc[78, 'c++'] = np.nan

#inspect data
print("===============inspected data============")
print(student.head())
print(student.info())
print(student.describe())
print(student.dtypes)
#df=student.drop_duplicates()
print(student.isnull().sum())

#cleaning data
print("=================cleaned data===============")
student['python']=student['python'].fillna(student['python'].mean())
student['java']=student['java'].fillna(student['java'].mean())
student['c++']=student['c++'].fillna(student['c++'].mean())
student['c']=student['c'].fillna(student['c'].mean())

print(student.isnull().sum())

print("=========== feature Engineering========")
subject_col=['python','java','c++','c']

student['Total']=student[subject_col].sum(axis=1)

student['percentage']=(student['Total']/400)*100

student['percentage']=student['percentage'].round(2)
print(student.tail())

print(student.describe)

print("=============group by===========")
group_data=student.groupby('subjects').agg({
    'python':['max','min','sum'],
    'java':['max','min','sum'],
    'c++': ['max','min','sum'],
    'c' : ['max','min','sum']
})
print(group_data)

print("==============pivot table=========")
pivot_data=pd.pivot_table(student,
                     values='Total',
                     index='subjects',
                     aggfunc='mean'     
)
print(pivot_data)

print("=============outliers========")
Q1=student['percentage'].quantile(0.25)
Q3=student['percentage'].quantile(0.75)
print("Q1 : ",Q1)
print("Q2 : ",Q3)

IQR=Q3-Q1
print("IQR : ",IQR)

lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

outliers=student[
    (student['percentage']<lower)|
    (student['percentage']>upper)
]
print(outliers)
Export_csv=student.describe()
print("======== export data========")
Export_csv.to_csv("student_analysis.csv")
```
