```python
#1. Create a Series using a list of 5 numbers.
import pandas as pd
list=pd.Series([10,20,30,40,50],index=['1','2','3','4','5'])
print(list)
```

```python
#2. Create a Series using a dictionary of student marks.
student=pd.Series({'dipali':90 ,'manas':88,'anu':99})
print(student)
```

```python
#3. Create a DataFrame using a list of dictionaries.
student_marks=pd.DataFrame({
    'Student_name': pd.Series(['dipali','manas','anu']),
    'Marks' : pd.Series([95,90,99])
})
print(student_marks)
```

```python
#4. Add a new column using scalar values.
student_marks['Class']='bcs'
print(student_marks)
```

```python
#5. Add a new column using a Series.
attendence=pd.Series([12,34,45])
student_marks['Attendence']=attendence
print(student_marks)
```

```python
#6. Rename one column and multiple columns.
student_marks.rename(columns={'Marks':'Student_marks'},inplace=True)
print(student_marks)

#rename Multiple Columns
student_marks.rename(columns={
    'Class':'Student_Class',
    'attendence':'student_attendence'
}, inplace=True)
print(student_marks)
```

```python
#7. Use loc to select:
#  - A single row
print(student_marks.loc[0])

# - Multiple rows
print(student_marks.loc[:, 'Student_Class'])

#- Conditional rows
print(student_marks.loc[student_marks['Student_marks']>90])
```

```python
#8. Use iloc to select:
#  - First row
print(student_marks.iloc[0])

#  - Second column
print(student_marks.iloc[:,1])

#  - First three rows and two columns
print(student_marks.iloc[:3,:1])
```

```python
#9. Find DataFrame shape and dtypes.
print("shape of the student marks : ",student_marks.shape)
print("dtype of student marks :",student_marks.dtypes)
```

```python
#10. Handle missing values using fillna
#adding nan values
import numpy as np
student_marks['student_class']=np.nan
student_marks['student_marks']=np.nan
#missing values
print("mising values: ",student_marks.isnull())
#fill null values
print("filling null values :",student_marks.fillna(0))
```
