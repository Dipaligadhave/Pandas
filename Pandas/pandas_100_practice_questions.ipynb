```python
### Q56. Detect outliers in a column
import numpy as np
Q1=dfnull1['marks'].quantile(0.25)
print("Q1 : ",Q1)

Q3=dfnull1['marks'].quantile(0.75)
print("Q3 : ",Q3)
IQR=Q3-Q1
print("IQR : ",IQR)

lower_outliers =Q1-1.5*IQR
print("lower_outliers  : ",lower_outliers)
upper_outliers=Q3+1.5*IQR
print("upper_outliers :",upper_outliers)
outliers=dfnull1[
    (dfnull1['marks']< lower_outliers)|
    (dfnull1['marks']> upper_outliers)
]
print("outliers : ",outliers)


print("========= handle outliers using clip=====================")
### Q57. Handle outliers using clipping.
print(dfnull1['marks'].clip(lower=lower_outliers, upper=upper_outliers))

print("==================another method using mean ans std==================")
# another method using mean ans std
out_mean=dfnull1['marks'].mean()
print("mean : ",out_mean)
out_std=dfnull1['marks'].std()
print("std : ", out_std)
z_score=(dfnull1['marks']-out_mean)/out_std
print("Z_score: ",z_score)
out=dfnull1['marks'][np.abs(z_score)>3]
print("out : ", out)

```

```python
#4. Find second largest element.
arr = np.array([[10, 25, 8],
                [45, 15, 30]])

#flat=arr.flatten()
print(np.sort(arr)[-2])
#Shortcut
second=np.sort(arr , axis=None)[-2]
print(second)

#2. Second Largest in the Entire 3D Array
arr = np.array([
    [[1, 8], [3, 4]],
    [[9, 6], [7, 2]]
])

secn=np.sort(arr , axis=None)[-2]
print(secn)
#if arr has duplicates
second_large=np.unique(arr)[-2]
print(second_large)

##4. Second Largest in Each Row
second_lar=np.sort(arr , axis=1)[:,-2]
print(second_lar)

##4. Second Largest in Each column
second_lar=np.sort(arr , axis=0)[:,-2]
print(second_lar)

#for better efficiency
lasrge=np.partition(arr , -2 , axis=None)[-2]
print(lasrge)
```

```python
#5. Find third largest element.
arr = np.array([[10, 25, 8,8],
                [45, 15, 30,30]])
flat=arr.flatten()
third_large=np.sort(flat)[-3]
print(third_large)

#6. Reverse an array.
rev=arr[::-1]
print(rev)

#7. Sort an array.
sor=np.sort(arr)
print(sor)

#8. Find unique elements.
print(np.unique(arr))

#9. Count frequency of each element.
uniqu , count=np.unique(arr , return_counts=True)
print("unique  :",uniqu)
print("count : ",count)

#Display as Element : Frequency
for i in range(len(uniqu)):
    print(uniqu[i], ":",count[i])

#Method 2: Using a Dictionary (Without np.unique())
arr = np.array([1, 2, 2, 3, 3, 3, 4])
freq={}
for i in arr : 
    if i in freq:
        freq[i]+=1
    else:
        freq[i]=1
print(freq)
```

```python
#Method 3: For a 2D Array
arr = np.array([[1, 2, 2],
                [3, 3, 4]])
uniqu , count=np.unique(arr , return_counts=True)
for u ,c in zip(uniqu, count):
    print(u , ":",c)
```
